# FinSight Portfolio Optimization Engine

## Risk-Based Portfolio Optimization Using Modern Portfolio Theory (MPT)

**Project Objective**

The objective of this project is to develop a portfolio optimization engine that accepts an investor's risk profile and investment capital and generates a suitable portfolio allocation.

The system analyzes historical stock data and uses Modern Portfolio Theory (MPT) to construct optimized portfolios for three investor profiles:

- Conservative
- Moderate
- Aggressive

The optimized portfolios are evaluated using:

- Expected Return

- Sharpe Ratio

```python
annual_return = average_daily_return * 252
annual_risk = daily_risk * (252 ** 0.5)

print("Annual Return:")
print(annual_return)

print("\nAnnual Risk:")
print(annual_risk)
```

```python
correlation = returns.corr()

correlation
```

```python
risk_free_rate = 0.06

risk_free_rate
```

```python
covariance_matrix = returns.cov() * 252

covariance_matrix
```

```python
import numpy as np

num_portfolios = 10000
num_stocks = len(stock_columns)

portfolio_results = []

for i in range(num_portfolios):

    weights = np.random.random(num_stocks)
    weights = weights / np.sum(weights)

    portfolio_return = np.dot(weights, annual_return)

    portfolio_risk = np.sqrt(
        np.dot(weights.T, np.dot(covariance_matrix, weights))
    )

    sharpe_ratio = (portfolio_return - risk_free_rate) / portfolio_risk

    portfolio_results.append([
        portfolio_return,
        portfolio_risk,
        sharpe_ratio,
        *weights
    ])

portfolio_results = pd.DataFrame(
    portfolio_results,
    columns=["return", "risk", "sharpe"] + stock_columns
)

portfolio_results.head()
```

```python
best_portfolio = portfolio_results.loc[
    portfolio_results["sharpe"].idxmax()
]

best_portfolio
```

```python
annual_return_with_cash = annual_return.copy()

annual_return_with_cash["CASH"] = risk_free_rate

annual_return_with_cash
```

```python
assets = stock_columns + ["CASH"]

assets
```

```python
cash_covariance = np.zeros((9, 9))

cash_covariance[:8, :8] = covariance_matrix.values

covariance_with_cash = pd.DataFrame(
    cash_covariance,
    index=assets,
    columns=assets
)

covariance_with_cash
```

```python
covariance_with_cash.loc["CASH"]
```

```python
def calculate_portfolio(weights):
    portfolio_return = np.dot(
        weights,
        annual_return_with_cash[assets]
    )

    portfolio_risk = np.sqrt(
        np.dot(
            weights.T,
            np.dot(covariance_with_cash.values, weights)
        )
    )

    sharpe_ratio = (
        portfolio_return - risk_free_rate
    ) / portfolio_risk

    return portfolio_return, portfolio_risk, sharpe_ratio
```

```python
conservative = risk[risk["risk_appetite"] == "conservative"].iloc[0]

min_equity = conservative["min_equity_pct"] / 100
max_equity = conservative["max_equity_pct"] / 100

min_equity, max_equity
```

```python
conservative_results = []

for i in range(10000):

    stock_weights = np.random.random(8)

    equity_weight = np.random.uniform(min_equity, max_equity)

    stock_weights = stock_weights / stock_weights.sum()
    stock_weights = stock_weights * equity_weight

    cash_weight = 1 - equity_weight

    weights = np.append(stock_weights, cash_weight)

    portfolio_return, portfolio_risk, sharpe = calculate_portfolio(weights)

    conservative_results.append([
        portfolio_return,
        portfolio_risk,
        sharpe,
        *weights
    ])

conservative_results = pd.DataFrame(
    conservative_results,
    columns=["return", "risk", "sharpe"] + assets
)

conservative_results.head()
```

```python
best_conservative = conservative_results.loc[
    conservative_results["sharpe"].idxmax()
]

best_conservative
```

```python
profile_results = {}

for profile_name in ["moderate", "aggressive"]:

    profile = risk[risk["risk_appetite"] == profile_name].iloc[0]

    min_equity = profile["min_equity_pct"] / 100
    max_equity = min(profile["max_equity_pct"] / 100, 1.0)

    results = []

    for i in range(10000):

        stock_weights = np.random.random(8)
        equity_weight = np.random.uniform(min_equity, max_equity)

        stock_weights = stock_weights / stock_weights.sum()
        stock_weights = stock_weights * equity_weight

        cash_weight = 1 - equity_weight
        weights = np.append(stock_weights, cash_weight)

        portfolio_return, portfolio_risk, sharpe = calculate_portfolio(weights)

        results.append([
            portfolio_return,
            portfolio_risk,
            sharpe,
            *weights
        ])

    results = pd.DataFrame(
        results,
        columns=["return", "risk", "sharpe"] + assets
    )

    profile_results[profile_name] = results

best_moderate = profile_results["moderate"].loc[
    profile_results["moderate"]["sharpe"].idxmax()
]

best_aggressive = profile_results["aggressive"].loc[
    profile_results["aggressive"]["sharpe"].idxmax()
]

print("MODERATE")
print(best_moderate)

print("\nAGGRESSIVE")
print(best_aggressive)
```

```python
final_portfolios = pd.DataFrame({
    "Profile": ["Conservative", "Moderate", "Aggressive"],
    "Expected Return": [
        best_conservative["return"],
        best_moderate["return"],
        best_aggressive["return"]
    ],
    "Risk": [
        best_conservative["risk"],
        best_moderate["risk"],
        best_aggressive["risk"]
    ],
    "Sharpe Ratio": [
        best_conservative["sharpe"],
        best_moderate["sharpe"],
        best_aggressive["sharpe"]
    ]
})

final_portfolios
```

# Final Project Summary

## Methodology

The portfolio optimization engine follows these steps:

1. Historical stock price data is loaded and checked for missing values.
2. Daily stock returns are calculated using percentage change.
3. Annualized expected returns and annualized risk are calculated.
4. The covariance matrix is calculated to measure relationships between stocks.
5. A 6% risk-free rate is used for Sharpe ratio calculation.
6. Multiple possible portfolios are generated using different asset weights.
7. Portfolio risk and expected return are calculated for each portfolio.
8. The Sharpe ratio is used to identify the portfolio with the best risk-adjusted return within each risk profile's equity allocation range.
9. The process is performed for Conservative, Moderate, and Aggressive investors.
10. The final portfolio weights are converted into actual investment amounts based on the investor's capital.

## Portfolio Evaluation Metrics

### Expected Return

Expected return represents the estimated annual return of the portfolio based on the historical return characteristics of the selected assets.

### Portfolio Risk

Portfolio risk represents the annualized volatility of the portfolio.

### Sharpe Ratio

The Sharpe ratio measures the return earned above the risk-free rate relative to the portfolio's risk.

**Sharpe Ratio = (Portfolio Return − Risk-Free Rate) / Portfolio Risk**

A higher Sharpe ratio indicates a higher return relative to the amount of risk taken.

```python
# Baseline Model

import pandas as pd
import numpy as np

# Load the portfolio dataset
portfolio = pd.read_csv("finsight_portfolio_data (1).csv")

# Stock columns
stock_columns = [
    "RELIANCE",
    "TCS",
    "HDFCBANK",
    "INFY",
    "ICICIBANK",
    "WIPRO",
    "BAJFINANCE",
    "HCLTECH"
]

# Calculate daily returns
returns = portfolio[stock_columns].pct_change().dropna()

# Calculate annual returns
annual_return = returns.mean() * 252

# Calculate annualized covariance
covariance_matrix = returns.cov() * 252

# Risk-free rate
risk_free_rate = 0.06

# Equal-weight portfolio
baseline_weights = np.ones(len(stock_columns)) / len(stock_columns)

# Portfolio return
baseline_return = np.dot(baseline_weights, annual_return)

# Portfolio risk
baseline_risk = np.sqrt(
    np.dot(
        baseline_weights.T,
        np.dot(covariance_matrix.values, baseline_weights)
    )
)

# Sharpe ratio
baseline_sharpe = (
    baseline_return - risk_free_rate
) / baseline_risk

# Display results
print("Baseline Portfolio Results")
print("Expected Return:", round(baseline_return * 100, 2), "%")
print("Risk:", round(baseline_risk * 100, 2), "%")
print("Sharpe Ratio:", round(baseline_sharpe, 2))
```
